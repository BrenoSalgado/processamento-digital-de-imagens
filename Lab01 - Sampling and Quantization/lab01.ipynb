1)	Utilizando imagem relogio.tif e faça uma reamostragem de seus pixels para 300, 150 e 72 dpi. Sabe-se que a imagem original possui 1250 dpi. Fazer os cálculos por meio do seu algortimo. Não usar o opencv para a conversão

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("relogio.tif", cv2.IMREAD_GRAYSCALE)

dpi_original = 1250
dpi_novo = 300

linhas, colunas = img.shape
escala = dpi_novo / dpi_original

novas_linhas = int(linhas * escala)
novas_colunas = int(colunas * escala)

img_300 = np.zeros((novas_linhas, novas_colunas), dtype=np.uint8)

for i in range(novas_linhas):
    for j in range(novas_colunas):
        i_orig = int(i / escala)
        j_orig = int(j / escala)
        img_300[i, j] = img[i_orig, j_orig]

plt.imshow(img_300, cmap='gray', vmin=0, vmax=255)
plt.title("Imagem reamostrada para 300 DPI")
plt.show()

cv2.imwrite("relogio_300dpi.tif", img_300)
print("300 DPI:", img_300.shape)
```

2)	Utilizando a imagem ctskull-256.tif, refaça a quantização de seus pixels utilizando de 7 a 1 bit(s) por pixel.  Fazer os cálculos por meio do seu algortimo. Não usar o opencv para a conversão

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# leitura da imagem
img = cv2.imread("ctskull-256.tif", cv2.IMREAD_GRAYSCALE)

linhas, colunas = img.shape

for bits in range(7, 0, -1):
    niveis = 2 ** bits
    passo = 256 / niveis

    img_q = np.zeros((linhas, colunas), dtype=np.uint8)

    for i in range(linhas):
        for j in range(colunas):
            nivel = int(img[i, j] / passo)
            img_q[i, j] = int(nivel * passo)

    plt.figure()
    plt.imshow(img_q, cmap='gray', vmin=0, vmax=255)
    plt.title(f"Quantização com {bits} bits")
    plt.show()

    cv2.imwrite(f"ctskull_{bits}bits.tif", img_q)

    print(f"{bits} bits -> {niveis} niveis")
```
